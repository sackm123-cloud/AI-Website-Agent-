```python
"""
AI STEM Automation Hub
main.py

Central controller for:
1. Daily AI/Robotics News
2. STEM Class Updates
3. Daily STEM Tip
4. Project Idea
5. Weekly Progress Report
6. Competition Update
"""

import argparse
import logging
from datetime import datetime

from agents.news_agent import run_news_agent
from agents.class_agent import run_class_agent
from agents.stem_tip_agent import run_stem_tip_agent
from agents.project_agent import run_project_agent
from agents.weekly_report_agent import run_weekly_report_agent
from agents.competition_agent import run_competition_agent


# ---------------------------------------------------------
# Logging
# ---------------------------------------------------------

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
)

logger = logging.getLogger("AI-STEM-Automation")


# ---------------------------------------------------------
# Agent Runner
# ---------------------------------------------------------

def run_agent(agent_name: str):
    """Run a selected automation agent."""

    logger.info("Starting agent: %s", agent_name)

    try:
        if agent_name == "news":
            result = run_news_agent()

        elif agent_name == "class":
            result = run_class_agent()

        elif agent_name == "tip":
            result = run_stem_tip_agent()

        elif agent_name == "project":
            result = run_project_agent()

        elif agent_name == "weekly":
            result = run_weekly_report_agent()

        elif agent_name == "competition":
            result = run_competition_agent()

        elif agent_name == "all":
            result = run_all_agents()

        else:
            raise ValueError(f"Unknown agent: {agent_name}")

        logger.info("Agent completed successfully: %s", agent_name)
        return result

    except Exception as exc:
        logger.exception(
            "Agent failed: %s | Error: %s",
            agent_name,
            exc,
        )
        return None


# ---------------------------------------------------------
# Run All Agents
# ---------------------------------------------------------

def run_all_agents():
    """Run all six AI automation agents."""

    logger.info("Running all AI STEM agents...")

    results = {}

    agents = [
        ("news", run_news_agent),
        ("class", run_class_agent),
        ("tip", run_stem_tip_agent),
        ("project", run_project_agent),
        ("weekly", run_weekly_report_agent),
        ("competition", run_competition_agent),
    ]

    for name, function in agents:

        try:
            logger.info("Running %s agent...", name)

            results[name] = function()

            logger.info("%s agent completed.", name)

        except Exception as exc:

            logger.exception(
                "%s agent failed: %s",
                name,
                exc,
            )

            results[name] = None

    logger.info("All agents finished.")

    return results


# ---------------------------------------------------------
# Application Information
# ---------------------------------------------------------

def show_status():
    """Display application status."""

    print()
    print("=" * 60)
    print("       AI STEM AUTOMATION HUB")
    print("=" * 60)
    print(f"Date: {datetime.now().strftime('%d-%m-%Y')}")
    print(f"Time: {datetime.now().strftime('%H:%M:%S')}")
    print()
    print("Available Agents:")
    print()
    print("  1. news         - Daily AI/Robotics News")
    print("  2. class        - STEM Class Update")
    print("  3. tip          - Daily STEM Tip")
    print("  4. project      - Project Idea")
    print("  5. weekly       - Weekly Progress Report")
    print("  6. competition  - Competition Update")
    print("  7. all          - Run all agents")
    print()
    print("=" * 60)
    print()


# ---------------------------------------------------------
# Command Line Interface
# ---------------------------------------------------------

def main():

    parser = argparse.ArgumentParser(
        description="AI STEM Automation Hub"
    )

    parser.add_argument(
        "agent",
        nargs="?",
        default="status",
        choices=[
            "news",
            "class",
            "tip",
            "project",
            "weekly",
            "competition",
            "all",
            "status",
        ],
        help="Agent to run",
    )

    args = parser.parse_args()

    if args.agent == "status":
        show_status()
        return

    run_agent(args.agent)


# ---------------------------------------------------------
# Entry Point
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```
